```python
#!/usr/bin/env python3
import re
import glob

def natural_sort_key(s):
    return [int(text) if text.isdigit() else text.lower() for text in re.split(r'(\d+)', s)]

def parse_qe_magnetization(file_pattern):
    files = sorted(glob.glob(file_pattern), key=natural_sort_key)
    
    print(f"{'File':<25} | {'Total Mag (Bohr mag)':<22} | {'Abs Mag (Bohr mag)':<20}")
    print("-" * 75)
    
    for f_path in files:
        tot_mag = "N/A"
        abs_mag = "N/A"
        try:
            with open(f_path, 'r') as f:
                content = f.read()
                
                # Match total magnetization
                tot_match = re.search(r"total\s+magnetization\s*=\s*([-+]?\d*\.\d+|\d+)", content, re.IGNORECASE)
                if tot_match:
                    tot_mag = tot_match.group(1)
                
                # Match absolute magnetization
                abs_match = re.search(r"absolute\s+magnetization\s*=\s*([-+]?\d*\.\d+|\d+)", content, re.IGNORECASE)
                if abs_match:
                    abs_mag = abs_match.group(1)
                    
            print(f"{f_path:<25} | {tot_mag:<22} | {abs_mag:<20}")
        except Exception as e:
            print(f"{f_path:<25} | Error reading file")

print("--- SCF ECUT FILES ---")
parse_qe_magnetization("scf_ecut_*.out")
print("\n--- SCF KPTS FILES ---")
parse_qe_magnetization("scf_kpts_*.out")
```
